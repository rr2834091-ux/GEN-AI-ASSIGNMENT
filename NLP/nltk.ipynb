```python
from bs4 import BeautifulSoup
import urllib.request
import nltk
data=urllib.request.urlopen('https://www.linkedin.com/feed')
content=data.read()
soup=BeautifulSoup(content)
content_data=soup.get_text(strip=True)
s_data=[i for i in content_data.split()]
fq=nltk.FreqDist(s_data)
for k,v in fq.items():
    print(str(k),':',str(v))
```

```python
fq.plot(10)
```

```python
data =urllib.request.urlopen('https://www.linkedin.com/feed')
content=data.read()
soup=BeautifulSoup(content)
text_data=soup.get_text(strip=True)
tkn=[i for i in text_data.split()]
clear_tkn=tkn[:]
word=stopwords.words('english')
for t in tkn:
    if t in stopwords.words('english'):
        clear_tkn.remove(t)
fd=nltk.FreqDist(clear_tkn)
for i,j in fd.items():
    print(str(i),'=',str(j))
fd.plot(10)
#Tokenizing the sentences:
from nltk.tokenize import sent_tokenize
data='''Thanks for wonderful opportunity.I am Ragul,From Chennai.I am a traveller and sports person.I love webseries.
interm of skills,I am a effective listner.Good in taking informed decisions.'''
print(sent_tokenize(data))
#Word tokenize:
from nltk.tokenize import word_tokenize
data1='''Thanks for wonderful opportunity.I am Ragul,From Chennai.I am a traveller and sports person.I love webseries.
interm of skills,I am a effective listner.Good in taking informed decisions.'''
print(word_tokenize(data1))
```

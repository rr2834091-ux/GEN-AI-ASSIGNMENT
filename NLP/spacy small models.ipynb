```python
import spacy
nlp=spacy.load('en_core_web_sm')
doc=nlp('A Generative AI Engineer designs, develops, and deploys AI applications that can generate text, images, code, audio, and other content using Large Language Models (LLMs).')
len(doc)
#etracting the words of pos:
print('Nouns')
print('-'*75)
nouns=[]
adjectives=[]
for i in doc:
  if i.pos_=='NOUN':
    nouns.append(i.text)
print(nouns)
print('-'*75)
#Adjectives:
print('Adjectives')
print('-'*25)
for i in doc:
    if i.pos_=='ADJ':
        adjectives.append(i.text)
print(adjectives)
print('-'*25)
#Random Pairing:
print('Random Pairing')
print('-'*23)
import random
print(random.choice(adjectives)+' '+random.choice(nouns))
print('-'*23)
#Verbs:
print('Verbs')
print('-'*45)
Verbs=[]
for i in doc:
    if i.pos_=='VERB':
        Verbs.append(i.text)
print(Verbs)
print('-'*45)
#Past Participles:
print('Past Participle')
print('-'*25)
pastparticple=[]
for i in doc:
    if i.tag_=='VBN':
        pastparticple.append(i.text)
print(pastparticple)
#Getting Large syntatic noun pharses:
print('Noun_chunks')
print('-'*25)
for i in doc.noun_chunks:
    print(i.text)
print('-'*25)
```

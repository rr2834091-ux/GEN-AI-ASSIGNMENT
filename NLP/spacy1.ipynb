```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc=nlp('Aspiring Generative AI Engineer with a strong foundation in Python, Machine Learning, and Natural Language Processing (NLP).')
len(doc)
doc.sents
# sentence segregation
for i in doc.sents:
    print(i.text)
# displaying lemma with word
for i in doc:
    print(i.text,'--',i.lemma_)
# lemmatizing individual sentence
sentence=list(doc.sents) # sentence is a list which converts the input paragraphs into series of sentence
print(sentence)
sentence=list(doc.sents)[0]
for i in sentence:
    print(i.text)
```
